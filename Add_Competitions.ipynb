```python
from statsbombpy import sb
import pandas as pd
import numpy as np
```

```python
import warnings

# Ignore specific warning
warnings.filterwarnings("ignore", category=UserWarning, module="statsbombpy.api_client")
```

```python
# max row and column
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

```python
comps = sb.competitions()
```

```python
comps
```

```python
comps = comps[comps.competition_gender == 'male'].reset_index(drop=True)
comps = comps.drop(index=[2, 18,19,20, 22,23,24, 25,28,29,30,31,32,33, 34, 52,53,54, 58,59, 66])
```

```python
comps.reset_index(drop=True, inplace=True)
```

```python
comps = comps.drop(index=[1,17,31,32,33,34,35,36,37,38,39,41,43])
```

```python
comps.reset_index(drop=True, inplace=True)
```

```python
comps
```

```python
all_shots = pd.DataFrame()
for i in comps.index:
    competition_id = comps['competition_id'][i]
    season_id = comps['season_id'][i]
    matches_ = sb.matches(competition_id = competition_id, season_id = season_id)
    print(f"Index: {i}, Comp_ID: {competition_id}, Match Count: {matches_.index.size}")
    for j in matches_.index:
        match_id = matches_['match_id'][j]
        events = sb.events(match_id = match_id)
        shots = events[events['type'] == 'Shot'].reset_index(drop=True)
        all_shots = pd.concat([all_shots, shots])
```

```python
all_shots.to_csv('all_shots_from_all_comps.csv', index=False)
```
