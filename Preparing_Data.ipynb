```python
import math

def calculate_angle(x, y):

  g0 = [120, 44]
  p = [x, y]
  g1 = [120, 36]

  v0 = np.array(g0) - np.array(p)
  v1 = np.array(g1) - np.array(p)

  angle = math.atan2(np.linalg.det([v0,v1]),np.dot(v0,v1))
  return(abs(np.degrees(angle)))

def calculate_distance(x, y):
  x_dist = 120-x
  y_dist = 0
  if (y<36):
    y_dist = 36-y
  elif (y>44):
    y_dist = y-44
  return math.sqrt(x_dist**2 + y_dist**2)
```

```python
def count_opponents_around_shot(freeze_frame_data, shot_x, shot_y, radius=1):
  """
  Counts the number of opponents within a given radius of the shot location.

  Args:
    freeze_frame_data (list): A list of dictionaries representing the freeze frame.
    shot_x (float): The x-coordinate of the shot.
    shot_y (float): The y-coordinate of the shot.
    radius (float): The radius around the shot to consider.

  Returns:
    int: The number of opponents within the specified radius.
  """
  opponent_count = 0
  for player in freeze_frame_data:
    # Check if the player is an opponent and not the goalkeeper
    if not player['teammate']:
      player_x, player_y = player['location']
      # Calculate the distance between the player and the shot location
      distance = np.sqrt((player_x - shot_x)**2 + (player_y - shot_y)**2)
      if distance <= radius:
        opponent_count += 1
  return opponent_count
```

```python
from shapely.geometry import Point, Polygon

def player_count_between_goal_line(freeze_frame_data, shot_x, shot_y):

    g0 = [120.1, 44.5]
    g1 = [120.1, 35.5]

    area = Polygon([g0, g1, [shot_x, shot_y - 0.5], [shot_x, shot_y + 0.5]])

    player_count = 0
    for player in freeze_frame_data:
        player_x, player_y = player['location']
        point = Point(player_x, player_y)
        if area.contains(point):
            player_count += 1

    return player_count
```

```python
def plot_polygon(shot):

    g0 = (120, 44.5)
    g1 = (120, 35.5)

    shot_x = shot['x']
    shot_y = shot['y']

    area = Polygon([g0, g1, [shot_x, shot_y - 0.5], [shot_x, shot_y + 0.5]])

    x, y = area.exterior.xy

    plt.figure(figsize=(6, 4))
    plt.plot(x, y, 'b-', label='Area Shape')
    plt.fill(x, y, color='lightblue', alpha=0.5)

    shot_point = [shot_x, shot_y]
    goal_line = [g0, g1]

    if shot_point:
        plt.scatter(*shot_point, color='green', label='Shot Point')

    if goal_line:
        plt.plot([goal_line[0][0], goal_line[1][0]], [goal_line[0][1], goal_line[1][1]],
                 'g--', label='Goal Line')

    freeze_frame_data = eval(shot['shot_freeze_frame'])

    for player in freeze_frame_data:
        player_x, player_y = player['location']
        if player_x > (shot_x - 3):
            player_color = 'blue' if player['teammate'] else 'red'
            plt.scatter(player_x, player_y, color=player_color)

    # şutun merkezde olduğu 1m yarıçaplı daire çiz
    circle = plt.Circle((shot_x, shot_y), 1, color='red', fill=False, linewidth=1)
    plt.gca().add_patch(circle)


    plt.grid(True)
    plt.legend()
    plt.xlabel('X Coordinate')
    plt.ylabel('Y Coordinate')
    plt.title('Polygon Visualization')


    ax = plt.gca()
    ax.set_aspect('equal', adjustable='box')  # 1:1 oranını korur
    ax.set_xlim(ax.get_xlim()[0]-5, 121)  # sadece sağ sınırı 120'ye sabitle

    # Y eksenini ters çevir
    plt.gca().invert_yaxis()

    plt.show()
```

```python
from shapely.geometry import Point, Polygon

#@title create_shot_freeze_frame

def create_shot_freeze_frame(shot):

    # Extract the freeze frame data, shot x, and shot y
    freeze_frame = eval(shot['shot_freeze_frame'])

    player_x = shot['x']
    player_y = shot['y']

    g0 = (120.1, 44.5)
    g1 = (120.1, 35.5)
    p0 = (float(player_x), float(player_y - 0.5))
    p1 = (float(player_x), float(player_y + 0.5))

    points = [g0, g1, p0, p1]
    area = Polygon(points)

    print("Polygon points:", points)


    # Calculate the player count in the shot angle
    player_count = player_count_between_goal_line(freeze_frame, player_x , player_y)

    # Polygon Visualization
    plot_polygon(shot)

    print(f"Number of players in the shot angle for the shot: {player_count}")
    print(f"Number of opponents around the shot: {count_opponents_around_shot(freeze_frame, player_x, player_y)}")


    pitch = VerticalPitch(pitch_type='statsbomb', pitch_color='grass', line_color='white', linewidth=2)

    # Create a new figure and axis
    fig, ax = plt.subplots(figsize=(10, 8))

    # Draw the pitch
    pitch.draw(ax=ax)

    # Set y-limits to show only the upper half
    ax.set_ylim(60, 125)


    # Plot players
    for player in freeze_frame:
        location = player['location']
        name = player['player']['name']
        color = 'blue' if player['teammate'] else 'red'
        color = 'white' if player['position']['name'] == 'Goalkeeper' else color

        # Mark players inside the area with small black dot
        point = Point(location)
        if area.contains(point):
            pitch.scatter(location[0], location[1], ax=ax, color='black', s=10, zorder=3)

        # Plot player normally
        pitch.scatter(location[0], location[1], ax=ax, color=color, s=100, edgecolors='black', linewidth=1.5)

    # Plot shooter
    pitch.scatter(player_x, player_y, ax=ax, color='purple', s=100, edgecolors='black', linewidth=2)


    plt.title("Shot Freeze Frame", fontsize=16)
    plt.show()
```

Based on the current dataframe and common football analytics practices, here are some features you could consider for calculating expected assists (xA) and expected goals on target (xGOT):

**Expected Assists (xA):**

These features would focus on the pass that leads to a shot:

*   **pass\_end\_location**: The coordinates where the pass ended. This is crucial as the location of the assist pass directly impacts the quality of the resulting shot.
*   **pass\_outcome**: Whether the pass was successful. Unsuccessful passes won't result in an assist opportunity.
*   **pass\_type**: Different pass types (e.g., cross, through ball) have different probabilities of leading to a shot and thus influencing xA.
*   **pass\_height**: The height of the pass can affect its difficulty to control and shoot from.
*   **pass\_technique**: The technique used for the pass might also play a role.
*   **pass\_cross**: A binary indicator if the pass was a cross. Crosses are often associated with assist opportunities.
*   **pass\_cut\_back**: A binary indicator if the pass was a cut-back. Cut-backs from the byline are also common sources of assists.
*   **pass\_switch**: A binary indicator if the pass was a switch of play.
*   **pass\_assisted\_shot\_id**: This links the pass to the shot, confirming it was an assist.
*   **possession\_team**: The team making the pass.
*   **player**: The player making the pass.
*   **minute**: The minute of the match the pass occurred in.
*   **play\_pattern**: The play pattern leading to the pass.
*   **under\_pressure**: Whether the passer was under pressure.

**Expected Goals on Target (xGOT):**

These features would focus on the characteristics of the shot once it is on target:

*   **shot\_end\_location**: The coordinates and height where the shot ended (if it was on target). This is a primary factor for xGOT as it describes where the ball was heading.
*   **shot\_body\_part**: The body part used to take the shot (foot, head).
*   **shot\_technique**: The technique used for the shot (e.g., Normal, Volley).
*   **shot\_first\_time**: Whether the shot was taken first time.
*   **shot\_one\_on\_one**: Whether the shot was a one-on-one situation.
*   **shot\_open\_goal**: Whether the shot was at an open goal.
*   **under\_pressure**: Whether the shooter was under pressure.
*   **shot\_deflected**: Whether the shot was deflected.
*   **shot\_saved\_to\_post**: Whether the shot was saved onto the post by the goalkeeper.
*   **shot\_saved\_off\_target**: Whether the shot was saved but went off target.
*   **location (of the shot)**: The x and y coordinates from where the shot was taken.
*   **angle**: The angle to the goal from the shot location.
*   **distance**: The distance to the goal from the shot location.
*   **num\_opponents\_around**: The number of opponents around the shooter.
*   **num\_blocking\_players**: The number of players between the shooter and the goal.

Remember that the specific features you use will depend on the data available and the model you choose to build. You may need to perform further feature engineering and selection to optimize your xA and xGOT models.
